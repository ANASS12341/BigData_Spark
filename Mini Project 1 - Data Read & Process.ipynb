```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()
df = spark.read.csv("/Volumes/workspace/default/shared_files/first_100_customers.csv", inferSchema=True, header=True)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
# Changing date format and change is_active to boolean
from pyspark.sql.functions import *
df = df.withColumn("registration_date",to_date( col("registration_date"),"yyyy-MM-dd") )\
    .withColumn("is_active", col("is_active").cast('boolean'))
```

```python
df.show(5)
```

```python
# Remlacer les valeur vide par unknown
df = df.fillna("unknown")
df.show(5)
```

```python
df = df.fillna({'city':'unknown','state':'unknown','country':'unknown'})
df.show()
```

```python
# get distinct cities country and state
df.select("city","country","state").distinct().show()
```

```python
# ajouter une colonne registration_year 
df = df.withColumn('registration_year',year(col('registration_date')))\
    .withColumn('registration_month',month(col('registration_date')))\
    .withColumn('registration_day',dayofmonth(col('registration_date')))
df.show()
```

```python
# count customers by city country state
df.groupBy("city","country","state").count().show()
```

```python
df.groupBy("registration_date").count().show()
```

```python
df.groupBy("city").count().orderBy(desc("count")).show()
```

```python
# Pivot Table - Count of Active and Inactive Users Per state
df.groupBy("state").pivot("is_active").count().show()
```
