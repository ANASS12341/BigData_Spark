```python
from pyspark.sql import SparkSession
```

```python
# Initialize Spark Session
spark = SparkSession.builder.appName("Handling_Data_Type_In_Spark").getOrCreate()
```

```python
#Sample data 
data = [
(1,"John Doe","Bangalore","2023-01-15","152.75","True"),
(2,"Jane Smith","Delhi","2023-05-20","89.50","False"),
(3,"Robert Brown","Mumbai","InvalideDate","200.00","True"),
(4,"Linda White","Kolkata","2023-02-29",None,"yes"),# Feb 29 invalid in 2023
(5,"Mike Green","Chennai","2023-08-10","NaN","1"),#NaN needs handling
(6,"Sarah Blue","Hyderabad","InvalideDate","300.40","No"),
]

# Define column names
columns=["id","name","city","date","amount","is_active"]

# Create DataFrame 
df = spark.createDataFrame(data,schema=columns)

#Show the DataFrame
df.show()
```

```python
df.printSchema()
```

```python
# how to handle Integer Columns 
```

```python
from pyspark.sql.functions import col
from pyspark.sql.types import StructField ,IntegerType,FloatType,StringType,StructType
df = df.withColumn('id',col('id').cast(IntegerType()))
```

```python
df.show()
```

```python
df.printSchema()
```

```python
# handling Strings Columns
```

```python
from pyspark.sql.functions import *
df = df.withColumn('name_upper',upper(df.name))
df.show()
```

```python
df.filter(df.city.startswith('B')).show()
```

```python
df = df.withColumn('amount',col('amount').cast(FloatType()))
df.printSchema()
```

```python
# filling null values 
```

```python
df_failled = df.fillna({'amount':0})
df_failled.show()
```
